```python
# Toronto Parking Tickets (Jan 2024)
# Top 20 Streets / Areas with the Most Tickets

#  Load Data 
df = pd.read_csv("Parking_Tags_Data_2024_1.csv")

# Identify top streets based on ticket aomunts 
top_streets = (
    df["location2"]
    .astype(str)
    .str.strip()
    .str.upper()             
    .replace("", pd.NA)
    .dropna()
    .value_counts()
    .head(20)
)
print("\nTop 20 Toronto Streets/Areas by Parking Tickets - January 2024")
print(top_streets)

#  Visualization
plt.figure(figsize=(10, 8))
top_streets.sort_values().plot(kind="barh", color="steelblue")

plt.title("Top 20 Toronto Streets/Areas by Parking Tickets — January 2024", fontsize=14)
plt.xlabel("Number of Tickets", fontsize=12)
plt.ylabel("Street / Area", fontsize=12)
plt.tight_layout()
plt.savefig("top_20_toronto_ticket_streets_jan2024.png", dpi=150)
plt.show()

```
